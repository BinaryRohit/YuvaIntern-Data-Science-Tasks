```python
# Import the tools we need for data and graphs
import pandas as pd
import seaborn as sns

# Load the built-in restaurant tips dataset
df = sns.load_dataset('tips')

# Show the first 5 rows of our data
print(df.head())
```

```python
import matplotlib.pyplot as plt

# Make a scatter plot of total bill and tip
sns.scatterplot(data=df, x="total_bill", y="tip")

# Add a simple title to our graph
plt.title("Total Bill vs Tip Amount")

# Show the graph on the screen
plt.show()
```

```python
# Make a bar chart to see how many tables were served on each day
sns.countplot(data=df, x="day")

# Add a simple title to our graph
plt.title("Number of Tables Served per Day")

# Show the graph on the screen
plt.show()
```

```python
# Make a box plot to see bill amounts on different days
sns.boxplot(data=df, x="day", y="total_bill")

# Add a simple title to our graph
plt.title("Spread of Total Bills by Day")

# Show the graph on the screen
plt.show()
```

```python
# Make a histogram to see the most common bill amounts
sns.histplot(data=df, x="total_bill", bins=15)

# Add a simple title to our graph
plt.title("Distribution of Total Bills")

# Show the graph on the screen
plt.show()
```

```python
# Make a bar chart showing average tips by time of day and gender
sns.barplot(data=df, x="time", y="tip", hue="sex")

# Add a simple title to our graph
plt.title("Average Tip at Lunch vs Dinner")

# Show the graph on the screen
plt.show()
```
